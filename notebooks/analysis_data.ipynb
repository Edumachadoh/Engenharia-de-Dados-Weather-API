```python
# objetivo do arquivo:olhar dados em formato de tabela
path = '../data/weather_data.json'
import pandas as pd
import json
```

```python
# adicionar colunas para a visualização
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

```python
# abrir o arquivo para que o pandas consiga ler os dados
with open(path) as f:
    data =json.load(f)

df = pd.json_normalize(data)
df.head()
```
